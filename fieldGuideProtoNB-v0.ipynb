```python
#The library to make HTTP requests
import requests
```

```python
BASE = "https://api.inaturalist.org/v1/observations"

def get_observations(param: dict):
    r = requests.get(BASE, params=params, timeout=15)
    r.raise_for_status() #raises error for HTTP errors
    return r.json()
```

```python
def safe_get(url, **kwargs):
    try:
        r = requests.get(url, timeout=15, **kwargs)
        r.raise_for_status()
        return r
    except requests.RequestException as e:
        print("request failed: ", e)
        return None
```

```python
def iter_observations(base_params: dict):
    page = 1
    while True:
        params = base_params | {"page" : page} #pipe merges dict, rh-precedence for duplicates
        data = get_observations(params)#fetches a dict
        results = data["results"]
        if not results:
            break
        yield from results #yield returns 1 item at a time from a generator iterating over results.
        page += 1
```
